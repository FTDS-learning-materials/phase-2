## B. New Database

```python
# Open connection

client = pymongo.MongoClient(URI, server_api=pymongo.server_api.ServerApi(version="1", strict=True, deprecation_errors=True))
```

```python
# Get all databases

client.list_database_names()
```

```python
# Create a new database

db_world_info = client['world_info']
```

```python
# Create a new collection

coll_country_info = db_world_info['country']
```

When you create an empty database or empy collection, MongoDB Atlas will not display them on the web interface. However, newly created databases or newly collections will appear once they contain at least one document.

```python
# Insert a new data

data = {'_id': 1, 'name': 'india', 'region': 'asia', 'capital_city': 'new delhi', 'population': 1463865525}
coll_country_info.insert_one(data)
```

```python
# Insert another data

data = {'_id': 2, 'name': 'china', 'region': 'asia', 'capital_city': 'beijing', 'population': 1416096094}
coll_country_info.insert_one(data)
```

## C. Update, Delete, and Query Documents

```python
# Update a document

coll_country_info.update_one(
    {'name': 'indonesia'},                              # Filter based on country name
    {'$set': {'capital_city': 'ibu kota nusantara'}}    # New data
)
```

```python
# Update multiple documents

coll_country_info.update_many(
    {'$or': [
        {'name': {'$eq': 'nigeria'}},
				{'name': {'$eq': 'ethiopia'}}]},               # Filter based on country name
    {'$set': {'region': 'africa_continent'}}           # New data
)
```

```python
# Delete a document

coll_country_info.delete_one({'region': 'africa_continent'})
```

`delete_one` only deletes the first 1 record that matches the given criteria.

```python
# Delete multiple documents

coll_country_info.delete_many({'region': 'africa_continent'})
```

```python
# Find one document

results = coll_country_info.find_one({'region': 'asia'})
results
```

```python
# Find multiple documents

results = coll_country_info.find({'region': 'asia'})
results
```

```python
# Print all results

for doc in results:
  print(doc)
```

## D. Other Syntax

```python
# Count all documents in a collection

count = coll_country_info.count_documents({})
print('Total Data : ', count)
```

```python
# Display all data

results = coll_country_info.find({})
for doc in results:
  print(doc)
```

```python
# Limit only 4 data

results = coll_country_info.find({}).limit(4)
for doc in results:
  print(doc)
```

```python
# Sort data based on `population` from smallest to largest

results = coll_country_info.find().sort('population', 1)        # 1 : ascending and -1 : descending
for doc in results:
  print(doc)
```

### C.1 - Basic Query

From the list of models shown above, choose the one you'd like to use. In this notebook, we will be using the `gemini-3.1-flash-lite-preview` model. Feel free to use a different model if you prefer something more powerful.

````python
# Define the model

MODEL_ID = 'gemini-3.1-flash-lite-preview'
````

Now, let's write our first code to generate text with Gemini! We'll use the `generate_content` method to call the Gemini API.

````python
# Call Gemini
response= client.models.generate_content(
  model=MODEL_ID,
  contents=['Explain how AI works.']
)

# Show the response
print(response.text)
````

You can use the `Markdown` module to format the output more cleanly.

````python
# Here we import `Markdown` from `IPython.display`
# In Google Colab, there's no need to install it via pip since it's already included
# However, if you want to use this in VSCode, you'll likely need to install it first with: `pip install IPython`
from IPython.display import Markdown

# Display the response output as Markdown
Markdown(response.text)
````

**By default, the model returns a response only after it has finished generating the entire text.**

To get a faster, more interactive experience, you don't have to wait for the full result. Instead, **you can use streaming to handle partial results as they arrive**.

*It's a bit like watching a video on YouTube — you don't wait for the whole video to download before you start watching, right? The video plays as it downloads.*

To enable streaming, we'll use a different method called `generate_content_stream`. Example:

````python
# Here we'll use `generate_content_stream` to enable streaming
response = client.models.generate_content_stream(
    model='gemini-3.1-flash-lite-preview',
    contents='Write a story about a magic backpack.'
)

# Instead of waiting for the full response, the result is returned in chunks.
# A chunk is a piece of partial data that you can read incrementally
for chunk in response:
    print(chunk.text)
    # This is just a separator to help visualize where one chunk ends and the next begins
    print("_" * 80)
````

### C.3 - From Document Input a.k.a Document Understanding

Google Gemini can also be used to read files or documents in various formats, including PDF.

To learn more about document processing with Gemini, you can check out the following reference ([Explore document processing capabilities with the Gemini API](https://ai.google.dev/gemini-api/docs/document-processing?hl=id&lang=python))

In short, we're going to try giving Gemini a PDF file as input, and then ask it to **provide a summary of the document**.

````python
# Import libraries
from google.genai import types
import httpx
import base64

# URL of the PDF file we want to process
doc_url = 'https://discovery.ucl.ac.uk/id/eprint/10089234/1/343019_3_art_0_py4t4l_convrt.pdf'  # Replace with the actual URL of your PDF

# Fetch the PDF data from the URL and encode it as a file
doc_data = httpx.get(doc_url).content

# Define the prompt
prompt = "Summarize this document"

# Send the request using `generate_content` — pay attention to the `contents` section
response = client.models.generate_content(
    model=MODEL_ID,

    # Notice how the content is structured differently here
    contents=[
        # We're providing the PDF as raw bytes data
        types.Part.from_bytes(
            data=doc_data,                # The actual PDF data
            mime_type='application/pdf',  # Telling Gemini: "Hey, this is a PDF file!"
        ),
        prompt  # And here's what we want Gemini to do with it
    ]
)

# Let's see what the response looks like in Markdown
Markdown(response.text)
````

## D. Chat Conversation

We've seen how Gemini handles different input types — text, images, PDFs, audio. But so far, it's been one response and done.

**What if we want a real conversation? A back-and-forth with the AI?** Good news: YES, we can!

*Docs: [Explore Chat Capabilites via Generate Content](https://ai.google.dev/api/generate-content#chat).*

The Gemini Python SDK supports multi-turn interactions, allowing users to engage in iterative exchanges to progressively refine answers or address complex, multi-step problems.

This functionality includes built-in conversation history management, while underlyingly relying on the `generate_content` method.

The primary interface for this feature is the `.chats.create()` method.

````python
# Start a chat session with our model
chat = client.chats.create(model=MODEL_ID)

# Send a message (instead of `generate_content`)
response = chat.send_message('In one sentence, explain how a computer works to a young child.')

# Display the response
Markdown(response.text)
````

````python
# Send a follow-up message in the same chat
response = chat.send_message('Explain how to create email automation')

# Display the response
Markdown(response.text)
````

Once you're done chatting with the AI, you might want to check:

> ***"What did we talk about earlier?"***

Well, with the Gemini SDK, you can do exactly that. You can view the history or chat history of your conversation.

````python
# Curious about what we've talked about so far? Let's check the chat history!
for message in chat.get_history():

  # Who said this? (user or model)
  print(f'role - {message.role}', end=": ")

  # And what did they say?
  print(message.parts[0].text)
````

As shown in the output above, there are two distinct roles in the conversation:

1. `user` : Represents the human participant who initiates the chat.

2. `model` : Represents the Gemini AI that generates responses.

## E. Using Parameters

````python
# Creating a role of the AI
role = 'You are an AI expert. Use an analogy about food in your answer'

# Configuring the content generation settings to control the randomness and creativity of the AI's responses
chat_config = types.GenerateContentConfig(
    system_instruction=role,
    max_output_tokens=300,
    temperature=0,
    top_p=0.95,
    top_k=20
)

# Specify the model
MODEL_ID = "gemini-3.1-flash-lite-preview"

# Sending a request to the model
response = client.models.generate_content(
    model=MODEL_ID,
    config=chat_config,
    contents='What is Machine Learning'
)

# Display the response
Markdown(response.text)
````

When we set `system_instruction`, we provide the model with additional context to steer the model's behavior based on our specific needs and use case. We can set `system_instruction` with many different ways, including:

- Defining a persona or role (for a chatbot, for example).
- Defining output format (Markdown, YAML, etc.).
- Defining output style and tone (e.g., verbosit.y, formality, and target reading level)
- Defining goals or rules for a task (e.g., return code snippets without further explanation).
- Providing additional context for commands (e.g., knowledge cutoffs).

> ***<font color='red'>Remember: We set instructions when initializing the model, and then those instructions persist across all interactions with the model.</font>***

## F. Function Call

Function call **allow us to define custom functions that the model can call** during the generation process. This can be useful for tasks that require specific actions or interactions with external systems.

````python
# Case : Room temperature bot

## Step 1 : Define the functions
def power_on():
  '''
  Turn the AC on
  '''

  print('ROOM-TEMP-BOT : the AC is turned on')

def power_off():
  '''
  Turn the AC off
  '''

  print('ROOM-TEMP-BOT : the AC is turned off')

def increase_temperature():
  '''
  Increase the room temperature so that it is not too cold
  '''
  print('ROOM-TEMP-BOT : the temperature has been raised')

def decrease_temperature():
  '''
  Decrease the room temperature to make it cooler
  '''
  print('ROOM-TEMP-BOT : the temperature has been lowered')


## Step 2 : Insert the functions into the same list
room_controls = [power_on, power_off, increase_temperature, decrease_temperature]


## Step 3 : Give the model instruction
instruction = '''
  You are a helpful lighting system bot. You can turn
  lights on and off, and you can set the color. Do not perform any
  other tasks.

  You are a bot in the room control system.
  You can turn the AC on, off, lower the temperature, and raise the temperature.
  Do not perform any other tasks beyond those specified.
  '''
````

````python
# Config the model
chat = client.chats.create(
  model=MODEL_ID,
  config={
    'tools': room_controls,
    'system_instruction': instruction
  }
)

# Send a message
response = chat.send_message('This room has not been used for a long time')
print(response.text)
print('')

# Send another message
response = chat.send_message('Why is this room so hot?')
print(response.text)
````

### G.3 - Chat with EXA

Next, we will add the same prompt with an EXA AI module, where the model will search for the latest data/information on the internet.

````python
# Connect to EXA

exa = Exa(api_key=EXA_API_KEY)
````

````python
# Step 1 : Define the instruction

exa_instruction = """
You are a helpful and knowledgeable AI research assistant.
Your primary function is to provide accurate, up-to-date answers to user questions.

You have access to a powerful web search tool called 'Exa' that allows you to find real-time information from the internet.

**Your instructions are as follows:**

1.  **Analyze the User's Query:** First, understand the user's question.
    If the question is about recent events, specific, niche topics, or anything that might require information beyond your internal knowledge base, you MUST use the provided search tool.

2.  **Use the Search Tool (`exa_search_and_contents`):**
    * To use the tool, you must generate a function call.
    * Formulate a clear and concise search `query` parameter that best captures the user's intent.
    * Do not just repeat the user's question as the query. Synthesize it into effective search keywords.

3.  **Process the Search Results:**
    * After calling the tool, you will receive a list of web page results with their content.
    * Carefully read and synthesize the information from the provided content to construct your final answer.
    * Do not just copy the content. Explain it in your own words.

4.  **Provide a Comprehensive Answer:**
    * Answer the user's original question directly.
    * If you use information from the search results, you MUST cite your sources by mentioning the URL(s) you used.
    * If the search tool does not return relevant information or you cannot find an answer, clearly state that you were unable to find the information on the web. Do not invent an answer.
"""
````

````python
# Step 2 : Define the function to call EXA API

def exa_search_and_contents(query: str):
  '''
  Searches the internet for up-to-date information using a query.

  Args:
    query: A clear and concise search query string.

  Returns:
    A formatted string of search results including title, URL, and content snippet.
  '''
  print(f"Executing Exa search for: '{query}'")
  try:
    # Call the Exa API
    search_results = exa.search_and_contents(
      query=query,
      type="auto",
      num_results=3,  # Get the top 3 results for a good summary
      text={"max_characters": 2000}  # Limit content length per result
    )

    # Format the output for clarity
    formatted_results = []
    for result in search_results.results:
      formatted_results.append(
        f"Title: {result.title}\n"
        f"URL: {result.url}\n"
        f"Content: {result.text}\n"
        "-----------------"
      )

    return "\n".join(formatted_results)

  except Exception as e:
    return f"An error occurred during the search: {e}"
````

````python
# Send message with EXA API
chat = client.chats.create(
    model=MODEL_ID,
    config={
      "tools": [exa_search_and_contents]
    }
)

response = chat.send_message('Who are the winners of the 2026 National Governance Awards?')

# Display the response
Markdown(response.text)
````

Based on the results above, it can be seen that the model is able to answer the questions asked.

````python
# View the history of how the last response is formed

def print_history(chat):
  for content in chat.get_history():
    display(Markdown("###" + content.role + ":"))
    for part in content.parts:
      if part.text:
        display(Markdown(part.text))
      if part.function_call:
        print("Function call: {", part.function_call, "}")
      if part.function_response:
        print("Function response: {", part.function_response, "}")
    print("-" * 80)

print_history(chat)
````
